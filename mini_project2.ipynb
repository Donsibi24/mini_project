```python
#outlier handling in sqft_lot
# Outlier handling using IQR method

q1 = house_df['sqft_lot'].quantile(0.25)
q2 = house_df['sqft_lot'].quantile(0.50)
q3 = house_df['sqft_lot'].quantile(0.75)

print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')

iqr = q3 - q1

up_lim = q3 + (1.5 * iqr)
low_lim = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

```python
house_df.loc[house_df['sqft_lot'] > up_lim,'sqft_lot'] = up_lim

house_df.loc[house_df['sqft_lot'] < low_lim,'sqft_lot'] = low_lim
```

```python
#outlier handling in sqft_above
# Outlier handling using IQR method

q1 = house_df['sqft_above'].quantile(0.25)
q2 = house_df['sqft_above'].quantile(0.50)
q3 = house_df['sqft_above'].quantile(0.75)

print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')

iqr = q3 - q1

up_lim = q3 + (1.5 * iqr)
low_lim = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

```python
house_df=pd.get_dummies(house_df,dtype='int',drop_first= True)
```

#splitting and scaling

```python
x= house_df.drop('price',axis=1)
y=house_df['price']
```

```python
x1=x.copy()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x1 = scaler.fit_transform(x1)
```

```python
x1
```

```python
#train - test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#scaled x train and x test
x_train1, x_test1, y_train, y_test = train_test_split(x1, y, test_size=0.2, random_state=42)
```

#linear Regression

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
model = lr.fit(x_train1,y_train)
y_pred = model.predict(x_test1)
#using MSR, r2_score matrices for model evaluation
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
print('mean_squared_error',mean_squared_error(y_test,y_pred))
print('r2_score',r2_score(y_test,y_pred))
print('mean_absolute_error',mean_absolute_error(y_test,y_pred))
```

#lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error


lasso = Lasso(alpha=10,max_iter=10000)
lasso_model = lasso.fit(x_train1, y_train)
y_pred_lasso = lasso_model.predict(x_test1)
mse = mean_squared_error(y_test, y_pred_lasso)
r2 = r2_score(y_test, y_pred_lasso)
mae = mean_absolute_error(y_test, y_pred_lasso)

print(f"Lasso Regression MSE: {mse}")
print(f"Lasso Regression R2 Score: {r2}")
print(f"Lasso Regression MAE: {mae}")
```
